```python
import pandas as pd
from bokeh.io import output_notebook
output_notebook()
```

```python
dataset = pd.read_csv("Datasets/automobiles.csv")
dataset.head()
```

```python
from bokeh.plotting import figure, show
```

```python
dataset['index'] = dataset.index
```

```python
p1 = figure(title="Car prices",
            x_axis_label="Car Index",
            y_axis_label="Price")

p1.scatter(dataset['index'], dataset['price'])
show(p1)
```

```python
grouped_df = dataset.groupby('make', as_index=False)['price'].mean()
```

```python
grouped_plot = figure(title="Car Manufacturer Mean Prices",
                      x_axis_label="Car Manufacturer",
                      y_axis_label="Mean Price",
                      x_range=grouped_df['make'])

grouped_plot.xaxis.major_label_orientation = "vertical"
grouped_plot.scatter(grouped_df['make'], grouped_df['price'])
show(grouped_plot)
```

```python
from bokeh.models import LinearColorMapper

color_mapper = LinearColorMapper(
    palette='Magma256',
    low=grouped_df['price'].min(),
    high=grouped_df['price'].max()
)
```

```python
final_plot = figure(title="Car Manufacturer Mean Prices",
                    x_axis_label="Car Manufacturer",
                    y_axis_label="Mean Price",
                    x_range=grouped_df['make'])

final_plot.scatter(
    grouped_df['make'],
    grouped_df['price'],
    size=15,
    color={'field': 'y', 'transform': color_mapper}
)

final_plot.xaxis.major_label_orientation = "vertical"

show(final_plot)
```
